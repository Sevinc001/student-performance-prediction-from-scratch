## Target variable

G3 is the final grade (0-20 scale). For this project I want a classification problem (pass/fail) instead of regression, since I'm building a logistic regression model. Going to treat G3 >= 10 as "pass" like the original paper does.

```python
df['pass'] = (df['G3'] >= 10).astype(int)
df['pass'].value_counts(normalize=True)
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['G3'], bins=20, kde=True)
plt.axvline(10, color='red', linestyle='--', label='pass/fail cutoff')
plt.title('Distribution of Final Grade (G3)')
plt.legend()
plt.show()
```
